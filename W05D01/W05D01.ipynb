```python
# custom iterator

class CountDown:
    def __init__(self, start):
        self.start = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.start <= 0:
            raise StopIteration

        value = self.start
        self.start -= 1
        return value
```

```python
counter_obj = CountDown(5)
```

```python
for i in counter_obj:
    print(i)
```

```python
def fibonacci(n):
    a, b = 0, 1

    for i in range(n):
        yield a
        a, b = b, a + b
```

```python
def greet_decorator(func):

    def wrapper():
        print('Before function call')
        func()
        print('After function call')

    return wrapper
```

```python
@greet_decorator
def say_hi():
    print('hi')


@greet_decorator
def say_hello():
    print('Hello')
```

```python
say_hi()
```

```python
say_hello()
```

```python
def login_required(func):
    def wrapper(is_logged_in):
        if is_logged_in:
            return func(is_logged_in)
        print('Please login first')

    return wrapper
```

```python
def dashboard(is_logged_in):
    print('Welcome to dashboard')
```

```python
dashboard(False)
```

```python
class CustomOpen:
    def __init__(self, file_name, mode):
        self.file_name = file_name
        self.mode = mode

    def __enter__(self):
        print('Opened a file')
        self.fp = open(self.file_name, self.mode)
        return self.fp

    def __exit__(self, exc_type, exc_value, traceback):
        print('Close the file')
        self.fp.close()
```

```python
with CustomOpen('sample2.txt', 'w') as fp:
    fp.write('mango is yellow in color')
print('i am here means I am outside context')
```

```python
lock_available = True
```

```python
class ByAcquiringLock:
    def __enter__(self):
        global lock_available
        print('Acquiring a lock')
        if lock_available is False:
            raise Exception('Lock already acquired')
        else:
            lock_available = False

    def __exit__(self, exc_type, exc_value, traceback):
        global lock_available
        print('Releasing the lock')
        lock_available = True
```

```python
with ByAcquiringLock():
    print('doing something with the lock')
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
```

```python
p1 = Point(4, 3)
p2 = Point(5, 6)
```

```python
p3 = p1 + p2
```
