```python
# Virtual env

1. You can install any number of python from online. like python 3.6, python 3.10 (all of them are base)
2. If you are working on multiple projects. you would not want to pollute/spam your base python with all the libraries.
2. Also there is a chance that you may need different version of the same library in 2 different projects.
numpy v1 mohitboxoffice_env
numpy v2 mohitai_env

3. pip install virtualenv
4. Create an env:- virtualenv name_of_env
5. Activate the env to work on it:- 

5.1) Windows (Command Prompt):
.name_of_env\Scripts\activate

5.2) Windows (PowerShell):
.name_of_env\Scripts\Activate.ps1

5.3) macOS / Linux: 
source name_of_env/bin/activate


source pymd110426_env/bin/activate

6. confirm by either round brackets and env name in your command prompt/terminal or write
which python or where python.
pip freeze :- shows you all the installed libraries in your environment

```

```python
If someone wants to share a project with someone else:-
1. Share the project code on github.
2. For the Python Env:- python version and all the library version like numpy==1.1, pandas=2.3,etc 

To share the 2nd things, there is a standard which is to create a requirements.txt file and add all the library version
inside it. Share the requirements.txt file as part of the project itself on github.


pip install -r requirements.txt
```
